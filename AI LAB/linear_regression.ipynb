```python

from sklearn.model_selection import train_test_split

X = df.drop("MedHouseVal", axis=1)
print(X[:5])


```

```python

Y = df["MedHouseVal"]
print(Y[:5])
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("Y_train shape:", Y_train.shape)
print("Y_test shape:", Y_test.shape)
```

```python

X_simple_train = X_train["MedInc"]
X_simple_test = X_test["MedInc"]

print("Training feature shape:", X_simple_train.shape)
print("Testing feature shape:", X_simple_test.shape)
```

```python
print("\nFirst 5 training values:")
```

```python
x_mean = X_simple_train.mean()
y_mean = Y_train.mean()

print("Mean of training MedInc:", x_mean)
print("Mean of training MedHouseVal:", y_mean)
```

```python
# Convert the training data into NumPy arrays
x = X_simple_train.to_numpy()
y = Y_train.to_numpy()

# Calculate deviations from their respective means
x_deviation = x - x_mean
y_deviation = y - y_mean

# Calculate the numerator and denominator
numerator = np.sum(x_deviation * y_deviation)
denominator = np.sum(x_deviation ** 2)

# Calculate the slope
b1 = numerator / denominator

print("Numerator:", numerator)
print("Denominator:", denominator)
print("Slope (b1):", b1)
```

```python
b0 = y_mean - b1 * x_mean

print("Slope (b1):", b1)
print("Intercept (b0):", b0)
```

```python
# Predict using our regression equation
Y_pred_simple = b0 + b1 * X_simple_test

print("First 10 actual values:")
print(Y_test.head(10).to_numpy())

print("\nFirst 10 predicted values:")
print(Y_pred_simple.head(10).to_numpy())
```

```python
# Convert actual and predicted values into NumPy arrays
y_actual = Y_test.to_numpy()
y_predicted = Y_pred_simple.to_numpy()

# Calculate prediction errors
errors = y_actual - y_predicted

# 1. Mean Absolute Error
mae = np.mean(np.abs(errors))

# 2. Mean Squared Error
mse = np.mean(errors ** 2)

# 3. Root Mean Squared Error
rmse = np.sqrt(mse)

# 4. R-squared
ss_res = np.sum(errors ** 2)
ss_tot = np.sum((y_actual - np.mean(y_actual)) ** 2)

r2 = 1 - (ss_res / ss_tot)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R-squared:", r2)
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(y_actual, y_predicted, alpha=0.2)

# Perfect prediction reference line
plt.plot(
    [y_actual.min(), y_actual.max()],
    [y_actual.min(), y_actual.max()],
    "r--"
)

plt.xlabel("Actual House Value")
plt.ylabel("Predicted House Value")
plt.title("Simple Linear Regression: Actual vs Predicted")
plt.show()
```

```python
# Convert the training features and target into NumPy arrays
X_multi_train = X_train.to_numpy()
X_multi_test = X_test.to_numpy()

y_multi_train = Y_train.to_numpy()
y_multi_test = Y_test.to_numpy()

print("Training feature matrix:", X_multi_train.shape)
print("Testing feature matrix:", X_multi_test.shape)
print("Training target:", y_multi_train.shape)
print("Testing target:", y_multi_test.shape)
```

```python
# Add a column of ones to the training and testing matrices
X_multi_train_bias = np.column_stack(
    (np.ones(X_multi_train.shape[0]), X_multi_train)
)

X_multi_test_bias = np.column_stack(
    (np.ones(X_multi_test.shape[0]), X_multi_test)
)

print("Training matrix with intercept:", X_multi_train_bias.shape)
print("Testing matrix with intercept:", X_multi_test_bias.shape)

print("\nFirst training row before adding ones:")
print(X_multi_train[0])

print("\nFirst training row after adding ones:")
print(X_multi_train_bias[0])
```

```python
# Step 1: Transpose the training matrix
X_transpose = X_multi_train_bias.T

# Step 2: Calculate X transpose multiplied by X
A = X_transpose @ X_multi_train_bias

# Step 3: Calculate X transpose multiplied by Y
B = X_transpose @ y_multi_train

print("Shape of X transpose:", X_transpose.shape)
print("Shape of A (X transpose @ X):", A.shape)
print("Shape of B (X transpose @ Y):", B.shape)
```

```python
# Solve (X transpose @ X) b = X transpose @ Y
coefficients = np.linalg.solve(A, B)

print("All coefficients:")
print(coefficients)

print("\nIntercept:", coefficients[0])

print("\nFeature coefficients:")
for feature, coefficient in zip(X.columns, coefficients[1:]):
    print(feature, ":", coefficient)
```

```python
# Predict house values for the test set
y_pred_multi = X_multi_test_bias @ coefficients

print("First 10 actual values:")
print(y_multi_test[:10])

print("\nFirst 10 predicted values:")
print(y_pred_multi[:10])
```

```python
# Calculate prediction errors
errors_multi = y_multi_test - y_pred_multi

# Mean Absolute Error
mae_multi = np.mean(np.abs(errors_multi))

# Mean Squared Error
mse_multi = np.mean(errors_multi ** 2)

# Root Mean Squared Error
rmse_multi = np.sqrt(mse_multi)

# R-squared
ss_res_multi = np.sum(errors_multi ** 2)
ss_tot_multi = np.sum(
    (y_multi_test - np.mean(y_multi_test)) ** 2
)

r2_multi = 1 - (ss_res_multi / ss_tot_multi)

print("Multiple Regression Results")
print("MAE:", mae_multi)
print("MSE:", mse_multi)
print("RMSE:", rmse_multi)
print("R-squared:", r2_multi)
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(y_multi_test, y_pred_multi, alpha=0.2)

plt.plot(
    [y_multi_test.min(), y_multi_test.max()],
    [y_multi_test.min(), y_multi_test.max()],
    "r--"
)

plt.xlabel("Actual House Value")
plt.ylabel("Predicted House Value")
plt.title("Multiple Linear Regression: Actual vs Predicted")
plt.show()
```

```python
# Calculate residuals
residuals = y_multi_test - y_pred_multi

print("First 10 residuals:")
print(residuals[:10])

print("\nMean residual:", np.mean(residuals))
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(y_pred_multi, residuals, alpha=0.2)
plt.axhline(y=0, color="red", linestyle="--")

plt.xlabel("Predicted House Value")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Residual Plot: Multiple Linear Regression")
plt.show()
```

### 9. Interpretation of Regression Coefficients

In simple linear regression, the coefficient of `MedInc` is approximately 0.4193, indicating that a one-unit increase in median income is associated with an increase of approximately $41,934 in predicted median house value.

In multiple linear regression, the `MedInc` coefficient is approximately 0.4487. Holding the other features constant, a one-unit increase in median income is associated with an increase of approximately $44,867 in predicted house value. `HouseAge` and `AveBedrms` have positive coefficients, while `AveRooms`, `Population`, `AveOccup`, `Latitude`, and `Longitude` have negative coefficients in the fitted model.

The intercept in the multiple regression model is approximately -37.0233, but it has limited practical meaning because all input features being zero is unrealistic. The coefficients represent conditional associations rather than causal effects. The difference between the simple and multiple regression coefficients also shows that estimated relationships can change when additional features are included.

### Performance Comparison Between Simple and Multiple Regression

The performance of simple and multiple linear regression models was evaluated using MAE, MSE, RMSE, and R² score.

The simple linear regression model, which uses only median income (`MedInc`), achieved an MAE of 0.6299, an MSE of 0.7091, an RMSE of 0.8421, and an R² score of 0.4589.

The multiple linear regression model, which uses all eight input features, achieved an MAE of 0.5332, an MSE of 0.5559, an RMSE of 0.7456, and an R² score of 0.5758.

Therefore, multiple linear regression performs better on the test dataset. It produces lower prediction errors and explains approximately 57.58% of the variation in median house values, compared with 45.89% for simple linear regression. This indicates that considering additional features helps improve house-value prediction, although some variation remains unexplained.
