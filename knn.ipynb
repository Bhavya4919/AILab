```python
import csv
```

```python
#Prostate_Cancer.csv
data = []
with open('Prostate_Cancer.csv', 'r') as file:
    reader = csv.reader(file)
    header = next(reader)
    indexTarget = 1
    print(header)
    for row in reader:
        features = []
        for i in range(len(row)):
            if(i != indexTarget):
                features.append(float(row[i]))
            else:
                label = row[indexTarget]
        data.append([features, label])
print(data)
```

```python
import random

def stratified_split(X, y, test_ratio=0.2):
    random.seed(42)

    # Step 1: Group record indices by class label
    class_indices = {}

    for i in range(len(y)):
        label = y[i]

        if label not in class_indices:
            class_indices[label] = []

        class_indices[label].append(i)

    train_indices = []
    test_indices = []

    # Step 2: Split each class separately
    for label in class_indices:
        indices = class_indices[label]

        random.shuffle(indices)

        test_size = round(len(indices) * test_ratio)

        test_indices.extend(indices[:test_size])
        train_indices.extend(indices[test_size:])

    #Shuffle the final train and test indices
    
    random.shuffle(train_indices)
    random.shuffle(test_indices)

    # Step 3: Build the datasets using the same indices
    X_train = []
    y_train = []
    X_test = []
    y_test = []

    for i in train_indices:
        X_train.append(X[i])
        y_train.append(y[i])

    for i in test_indices:
        X_test.append(X[i])
        y_test.append(y[i])

    

    return X_train, X_test, y_train, y_test
```

```python
X = []
y = []

for row in data:
    X.append(row[0])
    y.append(row[1])
```

```python
X_train, X_test, y_train, y_test = stratified_split(X, y)
```

```python
def calculate_mean(values):
    total = 0

    for value in values:
        total += value

    return total / len(values)
```

```python
def calculate_std(values, mean):
    squared_sum = 0

    for value in values:
        difference = value - mean
        squared_sum += difference * difference

    return (squared_sum / len(values)) ** 0.5
```

```python
def fit_scaler(X_train):
    means = []
    stds = []

    num_features = len(X_train[0])

    for j in range(num_features):

        # Collect values from column j
        column = []

        for i in range(len(X_train)):
            column.append(X_train[i][j])

        # Calculate mean and standard deviation
        mean = calculate_mean(column)
        std = calculate_std(column, mean)

        # Handle a constant feature
        if std == 0:
            std = 1

        means.append(mean)
        stds.append(std)

    return means, stds
```

```python
def transform_data(X, means, stds):
    scaled_X = []

    for row in X:
        scaled_row = []

        for j in range(len(row)):
            scaled_value = (row[j] - means[j]) / stds[j]
            scaled_row.append(scaled_value)

        scaled_X.append(scaled_row)

    return scaled_X
```

```python
means, stds = fit_scaler(X_train)

X_train_scaled = transform_data(X_train, means, stds)
X_test_scaled = transform_data(X_test, means, stds)

print("First original row:", X_train[0])
print("First scaled row:", X_train_scaled[0])
```

```python
def euclidean_distance(v1, v2):
    squared_sum = 0
    for i in range(len(v1)):
        difference = v1[i] - v2[i]
        squared_sum += difference * difference
    distance = squared_sum ** 0.5
    return distance
```

```python
# 5. PREDICT ONE DATA POINT

def predict_one(X_train, y_train, new_point, k):
    distances = []

    for i in range(len(X_train)):
        distance = euclidean_distance(X_train[i], new_point)
        distances.append([distance, y_train[i]])

    # Sort by distance
    distances.sort(key=lambda item: item[0])

    # Select the K nearest neighbors
    neighbors = distances[:k]

    votes = {}

    for neighbor in neighbors:
        label = neighbor[1]

        if label in votes:
            votes[label] += 1
        else:
            votes[label] = 1

    prediction = max(votes, key=votes.get)

    return prediction


```

```python
# 6. PREDICT MULTIPLE DATA POINTS

def predict(X_train, y_train, X_test, k):
    predictions = []

    for point in X_test:
        result = predict_one(X_train, y_train, point, k)
        predictions.append(result)

    return predictions

```

```python


# 7. CALCULATE ACCURACY

def calculate_accuracy(y_actual, y_predicted):
    correct = 0

    for i in range(len(y_actual)):
        if y_actual[i] == y_predicted[i]:
            correct += 1

    accuracy = correct / len(y_actual) * 100

    return accuracy
```

```python
def multiclass_confusion_matrix(y_true, y_pred):
    classes = []

    for label in y_true + y_pred:
        if label not in classes:
            classes.append(label)

    n = len(classes)
    matrix = [[0] * n for _ in range(n)]

    for actual, predicted in zip(y_true, y_pred):
        i = classes.index(actual)
        j = classes.index(predicted)
        matrix[i][j] += 1

    return classes, matrix
```

```python
def accuracy(y_true, y_pred):
    correct = 0

    for actual, predicted in zip(y_true, y_pred):
        if actual == predicted:
            correct += 1

    return correct / len(y_true) if y_true else 0
```

```python
def precision(tp, fp):
    return tp / (tp + fp) if (tp + fp) else 0


def recall(tp, fn):
    return tp / (tp + fn) if (tp + fn) else 0


def f1_score(p, r):
    return 2 * p * r / (p + r) if (p + r) else 0
```

```python
classes, matrix = multiclass_confusion_matrix(y_test, predictions)

print("Classes:", classes)
print("Confusion Matrix:")

for row in matrix:
    print(row)
```

```python
def multiclass_metrics(y_true, y_pred):
    classes = []

    # Get all unique classes
    for label in y_true + y_pred:
        if label not in classes:
            classes.append(label)

    total_p = 0
    total_r = 0
    total_f1 = 0

    for cls in classes:
        tp = fp = fn = 0

        for actual, predicted in zip(y_true, y_pred):
            if actual == cls and predicted == cls:
                tp += 1
            elif actual != cls and predicted == cls:
                fp += 1
            elif actual == cls and predicted != cls:
                fn += 1

        p = precision(tp, fp)
        r = recall(tp, fn)
        f1 = f1_score(p, r)

        print(cls, "Precision:", p,
              "Recall:", r, "F1:", f1)

        total_p += p
        total_r += r
        total_f1 += f1

    n = len(classes)

    return total_p / n, total_r / n, total_f1 / n
```

```python
p, r, f1 = multiclass_metrics(y_test, predictions)

print("Macro Precision:", p)
print("Macro Recall:", r)
print("Macro F1-score:", f1)
```
