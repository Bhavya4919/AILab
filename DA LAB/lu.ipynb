```python
import numpy as np
```

```python
def input_matrix():
    n = int(input("Enter the order of matrix: "))
    A = []
    print("Enter the matrix elements row by row:")
    for i in range(n):
        row = input().split()
        for j in range(n):
            row[j] = float(row[j])
        A.append(row)
    return A
```

```python
def input_vector(n):
    print("Enter the elements of vector b")
    b = input().split()
    for i in range(n):
        b[i] = float(b[i])
    return b
    
```

```python
def lu_decomposition(A):
    n = len(A)
    L = np.eye(n, dtype = float)
    U = np.array(A, dtype = float)
    for i in range(n):
        for j in range(i+1, n):
            m = U[j][i] / U[i][i]
            L[j][i] = m
            for k in range(i, n):
                U[j][k] = U[j][k] - m*U[i][k]
    return L,U
```

```python
def forward_substitution(L, b):
    n = len(b)
    y = np.zeros(n, dtype=float)
    for i in range(n):
        total = 0
        for j in range(i):
            total += L[i][j] * y[j]
        y[i] = (b[i] - total) / L[i][i]
    return y
```

```python
def backward_substitution(U, y):
    n = len(y)
    x = np.zeros(n, dtype=float)
    for i in range(n - 1, -1, -1):
        total = 0
        for j in range(i + 1, n):
            total += U[i][j] * x[j]
        x[i] = (y[i] - total) / U[i][i]
    return x
```

```python
def display_matrix(A, name):
    print("\n" + name + ":")
    for row in A:
        for value in row:
            print(f"{value:8.2f}", end=" ")
        print()
```

```python
def displayM(A):
    for row in A:
        print(row)
```

```python
A = input_matrix()
n = len(A)
b = input_vector(n)
L, U = lu_decomposition(A)
if L is not None:
    display_matrix(L, "Matrix L")
    display_matrix(U, "Matrix U")
    y = forward_substitution(L, b)
    print("\nVector y:")
    print(y)
    x = backward_substitution(U, y)
    print("\nSolution:")
    print(f"x = {x[0]:.2f}")
    print(f"y = {x[1]:.2f}")
    print(f"z = {x[2]:.2f}")
    LU = np.dot(L, U)
    display_matrix(LU, "Matrix LU")
    print("\nVerification of A = LU:")
    if np.allclose(A, LU):
        print("LU decomposition successful.")
    else:
        print("LU decomposition unsuccessful.")
```
