```python
import numpy as np
```

```python
def input_matrix():
    n = int(input("Enter the order of matrix: "))
    A = np.zeros((n, n), dtype=float)

    print("Enter the matrix elements row by row:")
    for i in range(n):
        values = list(map(float, input().split()))

        if len(values) != n:
            raise ValueError(f"Enter exactly {n} values in each row.")

        for j in range(n):
            A[i][j] = values[j]

    return A
```

```python
def input_vector(n):
    print("Enter the elements of vector b:")
    values = list(map(float, input().split()))

    if len(values) != n:
        raise ValueError(f"Enter exactly {n} elements.")

    return np.array(values, dtype=float)

```

```python
def cholesky_decomposition(A):
    n = len(A)
    L = np.zeros((n, n), dtype=float)

    # Check whether A is symmetric
    if not np.allclose(A, A.T):
        print("Cholesky decomposition requires a symmetric matrix.")
        return None

    for i in range(n):
        for j in range(i + 1):

            total = 0.0

            for k in range(j):
                total += L[i][k] * L[j][k]

            if i == j:
                value = A[i][i] - total

                if value <= 0:
                    print("Matrix is not positive definite.")
                    return None

                L[i][j] = np.sqrt(value)

            else:
                L[i][j] = (A[i][j] - total) / L[j][j]

    return L
```

```python
def forward_substitution(L, b):
    n = len(b)
    y = np.zeros(n, dtype=float)

    for i in range(n):
        total = 0.0

        for j in range(i):
            total += L[i][j] * y[j]

        y[i] = (b[i] - total) / L[i][i]

    return y
```

```python
def backward_substitution(U, y):
    n = len(y)
    x = np.zeros(n, dtype=float)

    for i in range(n - 1, -1, -1):
        total = 0.0

        for j in range(i + 1, n):
            total += U[i][j] * x[j]

        x[i] = (y[i] - total) / U[i][i]

    return x

```

```python
def display_matrix(A, name):
    print("\n" + name + ":")
    for row in A:
        for value in row:
            print(f"{value:8.2f}", end=" ")
        print()
```

```python
# Main program
A = input_matrix()
n = len(A)
b = input_vector(n)

# Step 1: Cholesky decomposition A = LL^T
L = cholesky_decomposition(A)

if L is not None:
    LT = L.T

    display_matrix(L, "Matrix L")
    display_matrix(LT, "Matrix L Transpose")

    # Verify A = LL^T
    LLT = L @ LT
    display_matrix(LLT, "Matrix LL^T")

    print("\nVerification of A = LL^T:")
    if np.allclose(A, LLT):
        print("Cholesky decomposition successful.")
    else:
        print("Cholesky decomposition unsuccessful.")

    # Step 2: Forward substitution Ly = b
    y = forward_substitution(L, b)

    print("\nIntermediate vector y:")
    print(y)

    # Step 3: Backward substitution L^T x = y
    x = backward_substitution(LT, y)

    print("\nSolution of Ax = b:")
    for i in range(n):
        print(f"x{i+1} = {x[i]:.4f}")

    # Step 4: Verify the solution
    print("\nVerification of Ax = b:")
    if np.allclose(A @ x, b):
        print("Solution verified successfully.")
    else:
        print("Solution verification failed.")
```
