```python
import csv
```

```python
A = []
with open('student_performance_dataset.csv', 'r') as file:
    reader = csv.reader(file)
    header = next(reader)
    print(header)
    features= ['daily_study_hours', 'attendance_percent', 'exam_score', 'study_efficiency']
    for row in reader:
        data = []
        for j in range(len(row)):
            if(header[j] in features):
                data.append(float(row[j]))
                            
        A.append(data)
print(len(A))
print(len(A[0]))
```

```python
import numpy as np
A = np.array(A)
```

```python
print(A[:5])
```

```python
a1 = A[:,0]
a2 = A[:,1]
a3 = A[:,2]
a4 = A[:,3]
print(a1[:5])
print(a2[:5])
print(a3[:5])
print(a4[:5])
```

```python
def calculate_norm(v):
    total = 0.0
    for x in v:
        total += x * x
    return total ** 0.5
```

```python
u1 = a1
q1 = u1 / calculate_norm(u1)
print(q1[:5])

r12 = np.dot(a2, q1)
u2 = a2 - (r12 * q1)
q2 = u2 / calculate_norm(u2)
print(q2[:5])

r13 = np.dot(a3, q1)
r23 = np.dot(a3, q2)
u3 = a3 - (r13 * q1) - (r23 * q2)
q3 = u3 / calculate_norm(u3)
print(q3[:5])

r14 = np.dot(a4, q1)
r24 = np.dot(a4, q2)
r34 = np.dot(a4, q3)
u4 = a4 - (r14 * q1) - (r24 * q2) - (r34 * q3)
q4 = u4 / calculate_norm(u4)
print(q4[:5])
```

```python
print(np.allclose(np.dot(q4,q2), 0))
```

```python
Q  = np.column_stack((q1,q2,q3,q4))
Q.shape
```

```python
r11 = np.dot(q1, a1)
r22 = np.dot(q2, a2)
r33 = np.dot(q3, a3)
r44 = np.dot(q4, a4)

R = np.array([[r11, r12, r13, r14],
     [0, r22, r23, r24],
     [0, 0, r33, r34],
     [0, 0, 0, r44]])

print(R)
```

```python
A_reconstructed = Q @ R
print(A_reconstructed)
```

```python
print(np.allclose(A, A_reconstructed))
```

```python
QQT = Q.T @ Q
```

```python
I = np.eye(len(Q[0]))
```

```python
np.allclose(QQT, I)
```

```python
errorMatrix = A - A_reconstructed 
print(errorMatrix)
total = 0.0
for i in range(len(errorMatrix)):
    for j in range(len(errorMatrix[0])):
        total += errorMatrix[i][j] ** 2
forb_norm = total ** 0.5
print(forb_norm)
np.allclose(forb_norm, 0)
```
